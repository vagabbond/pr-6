<h1>Практичне заняття 6. Обробка винятків</h1>

1. Реалізуйте обробку винятків наступного коду:

   a = 'hello'

   b = 100

   c = a + b

2. Введіть з екрану 2 цілих числа. Порахуйте результат ділення двох чисел, врахувавши ситуацію з можливим нульовим значенням знаменника

3. Cтворення коммітів
    1.	Створіть нову папку
    2.	Переконайтеся, що ви не знаходитесь всередині іншого репозиторію
    3.	Виконайте ініціалізацію нового репозиторію
    4.	Створіть 2 нових пустих файли
    5.	Зробіть комміт, який включає обидва файли
    6.	Додайте довільний текст в обидва файли
    7.	Зробіть комміт зі змінами першого файла
    8.	Зробіть комміт зі змінами другого файла
    9.	Внесіть зміни до обох файлів
    10.	Зробіть один комміт, який включає зміни обох файлів
    11.	Покажіть список коммітів репозиторію


4. Робота з гілками
    1.	Створіть нову папку
    2.	Переконайтеся, що ви не знаходитесь всередині іншого репозиторію
    3.	Виконайте ініціалізацію нового репозиторію
    4.	Створіть пустий файл і зробіть з ним комміт
    5.	Створіть 2 нові гілки
    6.	Перейдіть в першу гілку, додайте довільний текст у файл і зробіть комміт зі змінами
    7.	Перейдіть у другу гілку, додайте довільний текст у файл і зробіть комміт зі змінами
    8.	Створіть нову гілку від другої гілки і перейдіть до нової гілки
    9.	Змініть текст файлу на створіть комміт зі змінами
    10.	Покажіть список гілок репозиторію
    11.	Видаліть останню гілку 

5. Злиття гілок
    1.	Створіть сценарій прямого злиття гілок (fast forward merge)
    2.	Створіть сценарій злиття гілок зі створенням merge комміту
    3.	Створіть сценарій злиття гілок с merge конфліктом, вирішите конфлікт для остаточного злиття гілок

6. Додавання змін в GitHub репозиторій
    1.	Сторіть новий проект на GitHub
    2.	Склонуйте новий репозиторій за допомогою терміналу
    3.	Створіть нову гілку
    4.	Створіть 3 комміта в новій гілці. Додані файли повити містити текст
    5.	Додайте зміни в GitHub репозиторій
    6.	Запросіть у проект MPTCourseTeacher та одного зі своїх колег у якості коллабораторів
    7.	Створіть pull request в основну гілку та додайте двох коллабораторів у якості рев’юверів
    8.	Отримайте два підтвердження від рев’юверів 
    9.	Зробіть pull request merge
